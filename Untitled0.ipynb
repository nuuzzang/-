```python
import gradio as gr
from datetime import datetime, date, timedelta
import re
import statistics

# ============================================
# 월경 주기 예측 및 건강관리 챗봇
# 대상자: 만 15~49세 가임기 여성
# 방식: 규칙 기반 대화형 건강교육 챗봇
# ============================================

# 확인된 의학적 참고자료
SOURCES = {
    "주기": "https://www.acog.org/womens-health/faqs/your-first-period",
    "청소년": "https://www.acog.org/clinical/clinical-guidance/committee-opinion/articles/2015/12/menstruation-in-girls-and-adolescents-using-the-menstrual-cycle-as-a-vital-sign",
    "통증": "https://www.acog.org/womens-health/faqs/dysmenorrhea-painful-periods",
    "출혈": "https://www.acog.org/womens-health/faqs/abnormal-uterine-bleeding",
    "무월경": "https://www.acog.org/womens-health/faqs/amenorrhea-absence-of-periods"
}

NOTICE = (
    "\n\n※ 본 챗봇은 건강교육용이며 진단, 처방, "
    "의료진의 진료를 대신하지 않습니다."
)

MENU = """
🌸 무엇을 도와드릴까요?

아래의 버튼을 직접 누르시거나 번호/내용을 입력해 주세요.

1. 월경주기 계산 및 다음 예정일 예측
2. 월경통 관리 (0~10점)
3. 증상별 건강관리 (복통·두통·피로·부종)
4. 월경 건강교육 (정상주기·월경불순·위생관리)
5. 부정출혈 확인 및 병원 방문 안내
6. 월경주기 기록 확인

'메뉴'를 입력하면 언제든 돌아올 수 있어요.
"""

def initial_state():
    return {
        "age": None,
        "stage": "age",
        "dates": [],
        "bleeding": {},
        "pain": None
    }

def emergency(text):
    t = text.replace(" ", "")
    return any(x in t for x in [
        "실신", "기절", "의식잃", "호흡곤란",
        "극심한복통", "갑자기너무아파",
        "심한어지러움", "한시간마다생리대"
    ])

def emergency_message():
    return (
        "🚨 응급진료가 필요한 증상일 수 있어요.\n\n"
        "심한 출혈, 실신, 호흡곤란, 갑작스러운 극심한 복통 "
        "등은 즉시 응급진료를 받아야 해요.\n"
        "필요하다면 119에 도움을 요청하세요.\n\n"
        "참고: " + SOURCES["출혈"] + NOTICE
    )

# ============================================
# 1. 월경주기 계산과 기록
# ============================================

def calculate_cycle(text, state):
    found = re.findall(r"\d{4}-\d{2}-\d{2}", text)

    if not found:
        state["stage"] = "dates"
        return (
            "📅 최근 월경 시작일을 2개 이상 입력해 주세요.\n\n"
            "⚠️ 주기 계산 오류를 방지하기 위해, 중간에 빠진 월경 시작일 없이 연속된 기록을 입력해 주세요.\n\n"
            "예: 2026-08-01, 2026-08-29, 2026-09-26\n\n"
            "월경이 시작된 첫날을 입력해야 해요.\n"
            "기존 기록에 날짜를 추가할 수도 있어요."
        )

    try:
        new_dates = [
            datetime.strptime(x, "%Y-%m-%d").date()
            for x in found
        ]
    except ValueError:
        return "날짜를 확인해 주세요. 예: 2026-09-26"

    if any(x > date.today() for x in new_dates):
        return "미래 날짜는 기록할 수 없어요."

    # 현재 대화에서만 기록 유지
    existing = [
        date.fromisoformat(x) for x in state["dates"]
    ]
    all_dates = sorted(set(existing + new_dates))
    state["dates"] = [x.isoformat() for x in all_dates]
    state["stage"] = "menu"

    if len(all_dates) < 2:
        return (
            "월경 시작일 1개를 기록했어요.\n"
            "다른 월경 시작일을 하나 더 입력해 주세요.\n"
            "⚠️ 중간에 빠진 기록 없이 연속해서 기입해 주셔야 예측 정확도를 확인하는 지표가 정밀해집니다."
        )

    intervals = [
        (all_dates[i] - all_dates[i-1]).days
        for i in range(1, len(all_dates))
    ]

    if any(x < 15 or x > 90 for x in intervals):
        return (
            "⚠️ 기록은 유지했지만 주기 계산을 보류했어요.\n"
            "중간에 빠진 월경 시작일 없이 연속된 기록을 입력해 주셔야 올바른 계산이 가능해요.\n"
            "월경 시작일을 빠뜨리거나 잘못 입력했을 수 있으니 확인 후 다시 시도해 주세요.\n"
            "지속적으로 주기가 매우 짧거나 길다면 진료받으세요.\n"
            "기록을 수정하려면 '기록 초기화'를 입력하세요."
        )

    avg = round(statistics.mean(intervals))
    next_day = all_dates[-1] + timedelta(days=avg)
    variation = max(intervals) - min(intervals)

    result = (
        "🌸 월경주기 계산 결과\n\n"
        f"기록된 시작일: {len(all_dates)}개\n"
        f"주기별 간격: {', '.join(str(x)+'일' for x in intervals)}\n"
        f"평균 월경주기: {avg}일\n"
        f"다음 월경 예정일: {next_day.isoformat()}\n\n"
    )

    if variation <= 7:
        result += "기록된 주기 사이의 차이가 비교적 작아요.\n"
    else:
        result += (
            "주기 변동이 있어 예정일의 불확실성이 커요.\n"
        )

    if state["age"] is not None:
        upper = 45 if state["age"] < 18 else 35
        if avg < 21 or avg > upper:
            result += (
                "⚠️ 연령에 따른 일반적인 주기 범위에서 "
                "벗어날 수 있어요. 반복된다면 진료받으세요.\n"
            )

    if next_day < date.today():
        result += (
            "\n계산된 예정일이 이미 지났어요. "
            "최근 월경이 시작되었다면 날짜를 추가해 주세요. "
            "월경이 늦어지고 임신 가능성이 있다면 "
            "임신 여부를 확인하세요.\n"
        )

    result += (
        "\n✔ 앞으로도 시작일을 꾸준히 기록해 보세요.\n"
        "✔ 날짜 기록이 많으면 주기 변동을 파악하기 좋습니다.\n"
        "✔ 다만 예측 정확도가 반드시 완벽하게 높아지는 것은 아니며, 개인별 주기의 정확도를 객관적으로 확인하는 보조 지표가 됩니다.\n"
        "✔ 중간에 누락된 월경 기록이 있으면 주기 분석에 오류가 생길 수 있어요.\n"
        "✔ 예정일을 피임이나 임신 여부 판단에 사용하지 마세요.\n\n"
        "더 궁금한 점이 있으신가요?\n\n"
        "참고: " + SOURCES["주기"] + NOTICE
    )
    return result

def show_records(state):
    if not state["dates"]:
        return "저장된 월경 시작일이 없어요. 1번을 이용해 주세요."

    return (
        "📅 현재 대화의 월경 시작일 기록\n\n" +
        "\n".join(state["dates"]) +
        "\n\n새 날짜를 추가하려면 '월경주기'라고 입력하세요.\n"
        "기록을 지우려면 '기록 초기화'라고 입력하세요."
    )

# ============================================
# 2. 월경통 관리 및 공감
# ============================================

def pain_management(text, state):
    t = text.replace(" ", "")

    # 사용자가 메뉴에서 막 진입했거나 stage가 pain이 아닌 경우
    if state["stage"] != "pain":
        state["stage"] = "pain"
        return (
            "🌷 월경통 때문에 불편하시겠어요.\n\n"
            "통증이 얼마나 심한가요?\n"
            "0점: 통증 없음\n"
            "1~3점: 가벼운 통증\n"
            "4~6점: 중간 정도의 통증\n"
            "7~10점: 심한 통증\n\n"
            "0~10 사이의 숫자를 입력해 주세요."
        )

    # 이제 통증 입력 단계(state["stage"] == "pain")이므로 숫자를 분석하여 점수로 매칭
    match = re.search(r"(?<!\d)(10|[0-9])\s*점?", text)
    if not match:
        return "0부터 10 사이의 숫자로 다시 입력해 주세요."

    pain = int(match.group(1))
    state["pain"] = pain
    state["stage"] = "menu"

    if pain == 0:
        msg = (
            "현재 통증이 없군요.\n"
            "✔ 월경 시작일과 증상을 꾸준히 기록하세요.\n"
            "✔ 규칙적인 수면과 식사를 유지해 보세요."
        )
    elif pain <= 3:
        msg = (
            "가벼운 통증이 있군요. 불편할 수 있겠어요.\n"
            "✔ 아랫배나 허리에 온찜질하기\n"
            "✔ 가벼운 걷기나 스트레칭하기\n"
            "✔ 충분한 휴식 취하기\n"
            "✔ 통증의 시작 시점과 변화를 기록하기"
        )
    elif pain <= 6:
        msg = (
            "통증 때문에 힘드시겠어요.\n"
            "✔ 온찜질과 가벼운 활동 시도하기\n"
            "✔ 충분히 휴식하기\n"
            "✔ 진통제 사용은 약사나 의료진과 상담하기\n"
            "✔ 일상생활에 지장이 있거나 반복되면 진료받기"
        )
    else:
        msg = (
            "통증이 심해서 많이 힘드시겠어요.\n"
            "✔ 빠른 시일 내 산부인과 진료받기\n"
            "✔ 무리한 활동을 피하고 쉬기\n"
            "✔ 통증 위치와 지속 시간을 기록하기\n"
            "🚨 갑작스러운 극심한 통증, 실신, "
            "임신 가능성과 동반된 심한 복통은 응급진료받기"
        )

    return (
        f"🌸 월경통 {pain}점 관리 안내\n\n" +
        msg + "\n\n" +
        "더 궁금한 점이 있으신가요?\n\n" +
        "참고: " + SOURCES["통증"] + NOTICE
    )

# ============================================
# 3. 증상별 건강관리
# ============================================

SYMPTOMS = {
    "복통": (
        "복통 때문에 불편하시겠어요.\n"
        "✔ 아랫배 온찜질하기\n"
        "✔ 가벼운 스트레칭과 휴식하기\n"
        "✔ 통증의 위치·정도 기록하기\n"
        "✔ 심하거나 평소와 다른 통증은 진료받기"
    ),
    "두통": (
        "두통이 있어 힘드시겠어요.\n"
        "✔ 충분한 수면과 휴식 취하기\n"
        "✔ 규칙적으로 식사하고 적절히 수분 섭취하기\n"
        "✔ 두통 발생 시기 기록하기\n"
        "✔ 갑작스러운 극심한 두통이나 마비·시야 이상은 응급진료받기"
    ),
    "피로": (
        "피로감이 심하면 일상생활도 힘들 수 있어요.\n"
        "✔ 충분한 수면 취하기\n"
        "✔ 규칙적이고 균형 잡힌 식사하기\n"
        "✔ 철분이 포함된 음식 섭취하기\n"
        "✔ 심한 피로와 어지러움이 지속되면 빈혈 등 확인하기"
    ),
    "부종": (
        "몸이 부어 불편하시겠어요.\n"
        "✔ 가벼운 걷기와 스트레칭하기\n"
        "✔ 짠 음식 과다 섭취 줄이기\n"
        "✔ 충분한 휴식 취하기\n"
        "✔ 한쪽 다리만 갑자기 붓고 아프면 신속히 진료받기"
    )
}

def symptom_management(text):
    t = text.replace(" ", "")
    selected = []

    if any(x in t for x in ["복통", "배아파", "배가아파"]):
        selected.append("복통")
    if "두통" in t or "머리아파" in t:
        selected.append("두통")
    if "피로" in t or "피곤" in t:
        selected.append("피로")
    if any(x in t for x in ["부종", "붓기", "부어"]):
        selected.append("부종")

    if not selected:
        return (
            "어떤 증상이 있나요? 보기를 보고 알려주세요.\n\n"
            "복통 / 두통 / 피로 / 부종\n\n"
            "해당 단어를 입력해주시면 맞춤 관리를 제공합니다."
        )

    return (
        "\n\n".join(
            f"🌸 {name} 관리\n{SYMPTOMS[name]}"
            for name in selected
        ) +
        "\n\n증상이 심하거나 지속되면 의료진과 상담하세요.\n\n" +
        NOTICE
    )

# ============================================
# 4. 월경 건강교육
# ============================================

def health_education(text, state):
    t = text.replace(" ", "")
    state["stage"] = "menu"

    common_footer = (
        "\n\n💡 안내가 완료되었습니다. 대화 단계가 다시 '메인 메뉴' 상태로 돌아왔습니다.\n"
        "더 궁금한 점이 있으시다면 언제든 아래의 버튼을 누르거나 직접 단어를 입력해 주세요!\n"
    )

    if any(x in t for x in ["정상", "주기란", "주기교육"]):
        return (
            "🌸 정상 월경주기 교육\n\n"
            "월경주기는 월경이 시작된 첫날부터 "
            "다음 월경이 시작되기 전날까지예요.\n\n"
            "✔ 성인: 일반적으로 21~35일\n"
            "✔ 청소년: 일반적으로 21~45일\n"
            "✔ 월경 기간: 보통 2~7일\n\n"
            "개인마다 차이가 있으며 청소년은 "
            "월경을 시작한 초기 몇 년 동안 불규칙할 수 있어요.\n"
            "주기와 출혈량의 변화를 기록해 보세요."
            + common_footer + "\n"
            "참고: " + SOURCES["청소년"] + NOTICE
        )

    if "불순" in t or "불규칙" in t:
        return (
            "🌸 월경불순 건강교육\n\n"
            "월경이 불규칙해서 걱정되실 수 있어요.\n"
            "스트레스, 체중 변화, 과도한 운동, "
            "호르몬 변화, 일부 질환 등이 영향을 줄 수 있어요.\n\n"
            "✔ 월경 시작일 기록하기\n"
            "✔ 규칙적인 식사와 수면 유지하기\n"
            "✔ 무리한 다이어트 피하기\n"
            "✔ 불규칙한 주기가 지속되면 진료받기\n"
            "✔ 3개월 이상 월경이 없다면 진료받기\n"
            "✔ 임신 가능성이 있으면 임신 여부 확인하기"
            + common_footer + "\n"
            "참고: " + SOURCES["무월경"] + NOTICE
        )

    if "위생" in t or "생리대" in t:
        return (
            "🌸 월경 중 위생관리\n\n"
            "✔ 생리대는 출혈량에 맞춰 자주 교체하기\n"
            "✔ 교체 전후 손 씻기\n"
            "✔ 외음부는 부드럽게 씻기\n"
            "✔ 질 내부 세척 피하기\n"
            "✔ 가려움·악취·이상 분비물이 지속되면 진료받기\n\n"
            "깨끗한 위생 습관을 유지해 보세요."
            + common_footer + "\n" + NOTICE
        )

    if "운동" in t or "생활습관" in t:
        return (
            "🌸 월경 중 생활습관 관리\n\n"
            "✔ 규칙적인 수면과 식사\n"
            "✔ 몸 상태에 맞는 가벼운 운동\n"
            "✔ 충분한 휴식\n"
            "✔ 월경 시작일과 증상 기록\n\n"
            "작은 습관부터 꾸준히 실천해 보세요."
            + common_footer + "\n" + NOTICE
        )

    # 어떤 주제도 입력하지 않고 진입한 경우 교육 메뉴를 보여주고 사용자의 세부 입력(정상주기/불순/위생/생활습관)을 기다림
    state["stage"] = "education_detail"
    return (
        "🌸 어떤 건강정보가 궁금한가요?\n\n"
        "정상 월경주기 / 월경불순 / 위생관리 / 생활습관\n\n"
        "원하는 항목을 입력해 주세요."
    )

# ============================================
# 5. 부정출혈 단계별 질문 및 진료 안내
# ============================================

def bleeding_start(state):
    state["bleeding"] = {}
    state["stage"] = "bleeding_amount"
    return (
        "🩸 월경 기간이 아닌데 출혈이 있었군요.\n"
        "걱정되실 수 있어요. 몇 가지를 확인할게요.\n\n"
        "먼저 출혈량은 어느 정도인가요?\n"
        "1. 소량 (속옷에 조금 묻는 정도)\n"
        "2. 보통 (월경과 비슷한 양)\n"
        "3. 많음 (생리대가 빠르게 흠뻑 젖음)\n\n"
        "1~3번 중 선택해 주세요."
    )

def bleeding_step(text, state):
    stage = state["stage"]
    t = text.replace(" ", "")

    if stage == "bleeding_amount":
        if t in ["1", "소량"]:
            state["bleeding"]["amount"] = "소량"
        elif t in ["2", "보통"]:
            state["bleeding"]["amount"] = "보통"
        elif t in ["3", "많음", "과다"]:
            state["bleeding"]["amount"] = "많음"
        else:
            return "출혈량을 1. 소량 / 2. 보통 / 3. 많음 중 선택해 주세요."

        state["stage"] = "bleeding_duration"
        return (
            "출혈이 며칠 동안 지속되었나요?\n"
            "예: 2일 또는 5일\n\n"
            "심한 어지러움이나 실신, 호흡곤란이 있으면 "
            "질문을 계속하지 말고 즉시 응급진료를 받으세요."
        )

    if stage == "bleeding_duration":
        match = re.search(r"\d+", t)
        if not match:
            return "출혈이 지속된 기간을 숫자로 알려주세요. 예: 3일"

        days = int(match.group())
        if days < 1 or days > 365:
            return "출혈 기간을 다시 확인해 주세요."

        state["bleeding"]["duration"] = days
        state["stage"] = "bleeding_symptoms"
        return (
            "출혈과 함께 나타나는 증상이 있나요?\n\n"
            "1. 없음\n"
            "2. 복통\n"
            "3. 어지러움\n"
            "4. 심한 복통 또는 실신\n\n"
            "해당 번호를 입력해 주세요."
        )

    if stage == "bleeding_symptoms":
        options = {
            "1": "없음", "2": "복통",
            "3": "어지러움", "4": "심한 증상"
        }
        if t not in options:
            return "동반 증상을 1~4번 중 선택해 주세요."

        state["bleeding"]["symptoms"] = options[t]

        if t == "4":
            state["stage"] = "menu"
            return emergency_message()

        state["stage"] = "bleeding_pregnancy"
        return (
            "마지막으로 임신 가능성이 있나요?\n\n"
            "1. 있음\n"
            "2. 없음\n"
            "3. 잘 모르겠음\n\n"
            "1~3번 중 선택해 주세요."
        )

    if stage == "bleeding_pregnancy":
        options = {
            "1": "있음", "2": "없음", "3": "모름"
        }
        if t not in options:
            return "1. 있음 / 2. 없음 / 3. 잘 모르겠음 중 선택해 주세요."

        state["bleeding"]["pregnancy"] = options[t]
        b = state["bleeding"]
        state["stage"] = "menu"

        summary = (
            "🩸 부정출혈 확인 결과\n\n"
            f"출혈량: {b['amount']}\n"
            f"지속 기간: {b['duration']}일\n"
            f"동반 증상: {b['symptoms']}\n"
            f"임신 가능성: {b['pregnancy']}\n\n"
        )

        if b["amount"] == "많음":
            advice = (
                "⚠️ 출혈량이 많아 신속한 진료가 필요해요.\n"
                "생리대가 1시간마다 흠뻑 젖는 출혈이 "
                "2시간 이상 반복되거나 어지러움·호흡곤란이 "
                "동반되면 즉시 응급진료를 받으세요."
            )
        elif b["symptoms"] == "어지러움":
            advice = (
                "⚠️ 출혈과 어지러움이 함께 있어요.\n"
                "빠르게 진료받으세요. 심한 어지러움이나 "
                "실신감이 있다면 응급진료를 받으세요."
            )
        elif b["pregnancy"] != "없음":
            advice = (
                "⚠️ 임신 가능성을 확인해야 해요.\n"
                "임신검사를 고려하고 신속하게 진료받으세요.\n"
                "특히 복통이 동반되거나 심해지면 "
                "응급평가가 필요할 수 있어요."
            )
        elif b["symptoms"] == "복통":
            advice = (
                "⚠️ 복통이 동반된 부정출혈이에요.\n"
                "빠른 시일 내 산부인과 진료를 권장해요.\n"
                "갑자기 통증이 심해지면 즉시 진료받으세요.\n"
            )
        elif b["duration"] >= 7:
            advice = (
                "⚠️ 출혈이 여러 날 지속되고 있어요.\n"
                "산부인과 진료를 통해 원인을 확인하세요."
            )
        else:
            advice = (
                "소량의 출혈이라도 다양한 원인이 있을 수 있어요.\n"
                "새롭게 발생했거나 반복된다면 "
                "산부인과 진료를 권장해요.\n"
            )

        return (
            summary + advice + "\n\n" +
            "✔ 출혈 날짜·양·동반 증상을 기록하세요.\n" +
            "✔ 챗봇은 출혈 원인을 진단하지 않아요.\n\n" +
            "참고: " + SOURCES["출혈"] + NOTICE
        )

# ============================================
# 6. 전체 대화 흐름
# ============================================

def chatbot_reply(message, history, state):
    if state is None:
        state = initial_state()

    text = message.strip()
    t = text.replace(" ", "")

    if not text:
        return "질문이나 번호를 입력해 주세요.", state

    # 응급 증상은 다른 질문보다 우선
    if emergency(text):
        state["stage"] = "menu"
        return emergency_message(), state

    if t in ["처음부터", "전체초기화"]:
        state = initial_state()
        return "처음부터 시작할게요. 만 나이를 입력해 주세요.", state

    if t in ["메뉴", "취소", "돌아가기"]:
        if state["age"] is None:
            state["stage"] = "age"
            return "먼저 만 나이를 입력해 주세요.", state
        state["stage"] = "menu"
        return MENU, state

    # 대상자 연령 확인
    if state["stage"] == "age":
        match = re.fullmatch(r"(?:만\s*)?(\d{1,3})\s*세?", text)
        if not match:
            return "안녕하세요! 먼저 만 나이를 숫자로 알려주세요. 예: 20", state

        age = int(match.group(1))
        if age < 15 or age > 49:
            return (
                "본 챗봇의 설계 대상은 만 15~49세예요.\n"
                "해당 연령이 아니라면 개인 상황에 맞는 "
                "의료진 상담을 권장해요.\n"
                "다시 입력하려면 만 나이를 알려주세요.\n"
            ), state

        state["age"] = age
        state["stage"] = "menu"
        return (
            f"안녕하세요! 만 {age}세로 확인했어요.\n"
            "특정 질환이 없어도 이용할 수 있어요.\n\n" +
            MENU
        ), state

    # 단계별 부정출혈 상담
    if state["stage"].startswith("bleeding_"):
        return bleeding_step(text, state), state

    # 월경 날짜 입력 대기
    if state["stage"] == "dates":
        return calculate_cycle(text, state), state

    # 통증 점수 입력 대기
    if state["stage"] == "pain":
        return pain_management(text, state), state

    # 세부 교육 내용 입력 대기 상태
    if state["stage"] == "education_detail":
        return health_education(text, state), state

    if "기록초기화" in t:
        state["dates"] = []
        state["stage"] = "menu"
        return "월경 시작일 기록을 초기화했어요.", state

    if t in ["6", "기록확인", "월경기록"]:
        return show_records(state), state

    if t in ["1", "월경주기", "생리주기"] or any(
        x in t for x in ["월경주기계산", "생리주기계산", "예정일"]
    ):
        return calculate_cycle(text, state), state

    # 메뉴 판에서 2번 입력 혹은 월경통 단어 입력 시 처리
    if t in ["2", "월경통", "생리통", "월경통관리"] or any(
        x in t for x in ["통증", "생리통", "월경통"]
    ):
        return pain_management(text, state), state

    if t in ["3", "증상", "증상관리"]:
        return symptom_management(""), state

    if any(x in t for x in [
        "복통", "두통", "피로", "피곤", "부종",
        "붓기", "배아파", "머리아파"
    ]):
        return symptom_management(text), state

    if t in ["4", "건강교육", "월경건강교육"]:
        return health_education("", state), state

    if any(x in t for x in [
        "정상주기", "월경불순", "불규칙",
        "위생", "생리대", "생활습관", "운동"
    ]):
        return health_education(text, state), state

    if t in ["5", "부정출혈", "출혈", "병원방문"]:
        return bleeding_start(state), state

    if "과다출혈" in t or "출혈많음" in t:
        return (
            "⚠️ 출혈량이 많다면 신속히 진료받으세요.\n"
            "생리대가 매시간 흠뻑 젖는 출혈이 반복되거나 "
            "어지러움이 있다면 응급진료를 받으세요.\n\n"
            + bleeding_start(state)
        ), state

    if "병원" in t or "진료" in t:
        return (
            "🏥 병원 방문 안내\n\n"
            "✔ 일상생활이 어려운 심한 월경통\n"
            "✔ 반복되는 부정출혈\n"
            "✔ 지속적인 월경불순\n"
            "✔ 과다출혈 또는 장기간 출혈\n"
            "✔ 원인을 알 수 없는 3개월 이상의 무월경\n\n"
            + NOTICE
        ), state

    if any(x in t for x in [
        "걱정", "불안", "무서워", "힘들어", "속상"
    ]):
        return (
            "월경 때문에 걱정되고 힘드시겠어요. "
            "그런 마음이 들 수 있어요.\n\n"
            "어떤 점이 가장 걱정되나요?\n"
            "월경주기, 통증, 부정출혈 중 하나를 "
            "말씀해 주시면 건강정보를 안내해 드릴게요.\n"
            "증상이 심하거나 지속된다면 진료받으세요."
        ), state

    return (
        "말씀하신 내용을 정확히 이해하지 못했어요.\n"
        "아래 메뉴에서 원하는 기능을 선택해 주세요.\n"
        + MENU
    ), state

# ============================================
# 7. Gradio 대화형 화면
# ============================================

with gr.Blocks(title="월경 주기 예측 및 건강관리 챗봇") as demo:
    gr.Markdown(
        "# 🌸 월경 주기 예측 및 건강관리 챗봇\n"
        "만 15~49세 가임기 여성을 위한 "
        "월경주기 예측 및 건강교육 챗봇입니다.\n\n"
        "**안내:** 건강교육용이며 의료진의 진료를 "
        "대체하지 않습니다.\n"
        "- 세부 질문이나 교육 과정 중 언제든지 **'메뉴'**를 입력하시면 홈 화면으로 올 수 있습니다."
    )

    chat = gr.Chatbot(
        value=[{
            "role": "assistant",
            "content": "안녕하세요! 월경 주기 예측 및 건강관리 챗봇입니다.\n\n먼저 만 나이를 알려주세요. 예: 20"
        }],
        height=400,
        label="챗봇 대화"
    )

    with gr.Row():
        btn1 = gr.Button("📅 1. 주기 계산 / 예정일 예측")
        btn2 = gr.Button("🌷 2. 월경통 관리")
        btn3 = gr.Button("🤒 3. 증상별 건강관리")
    with gr.Row():
        btn4 = gr.Button("📚 4. 월경 건강교육")
        btn5 = gr.Button("🩸 5. 부정출혈 / 병원 안내")
        btn6 = gr.Button("📋 6. 나의 월경기록 확인")

    user_box = gr.Textbox(
        label="나 (직접 입력도 가능해요)",
        placeholder="메시지를 입력하거나 위의 메뉴 버튼을 직접 클릭해 주세요!",
        lines=2
    )

    with gr.Row():
        send = gr.Button("전송", variant="primary")
        reset = gr.Button("대화 및 기록 초기화")

    state = gr.State(initial_state())

    def send_message(message, history, state):
        if not message.strip():
            return "", history, state

        if history is None:
            history = []

        reply, state = chatbot_reply(message, history, state)

        history = history + [
            {"role": "user", "content": message},
            {"role": "assistant", "content": reply}
        ]

        return "", history, state

    # 버튼 클릭 시 텍스트 전송과 동일한 기능 수행하도록 바인딩
    btn1.click(lambda: "1", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )
    btn2.click(lambda: "2", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )
    btn3.click(lambda: "3", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )
    btn4.click(lambda: "4", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )
    btn5.click(lambda: "5", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )
    btn6.click(lambda: "6", outputs=user_box).then(
        send_message, inputs=[user_box, chat, state], outputs=[user_box, chat, state]
    )

    send.click(
        send_message,
        inputs=[user_box, chat, state],
        outputs=[user_box, chat, state]
    )

    user_box.submit(
        send_message,
        inputs=[user_box, chat, state],
        outputs=[user_box, chat, state]
    )

    def reset_all():
        return (
            [{
                "role": "assistant",
                "content": "안녕하세요! 먼저 만 나이를 알려주세요."
            }],
            initial_state(),
            ""
        )

    reset.click(
        reset_all,
        outputs=[chat, state, user_box]
    )

    gr.Markdown(
        "### 의학적 참고자료\n"
        "- [ACOG 월경주기 교육](" + SOURCES["주기"] + ")\n"
        "- [ACOG 청소년 월경주기](" + SOURCES["청소년"] + ")\n"
        "- [ACOG 월경통 관리](" + SOURCES["통증"] + ")\n"
        "- [ACOG 비정상 자궁출혈](" + SOURCES["출혈"] + ")\n\n"
        "※ 입력 내용은 현재 챗봇 세션에서만 활용하도록 "
        "설계했으며 별도의 파일이나 데이터베이스에 "
        "저장하는 코드는 포함하지 않았습니다.\n"
    )

demo.launch()
```
